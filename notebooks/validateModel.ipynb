```python
import boto3
import json

REGION = "ap-southeast-2"
ENDPOINT_NAME = "customer-churn-new-endpoint"

aws_session = boto3.Session(region_name=REGION)
runtime = aws_session.client(
    "sagemaker-runtime",
    region_name=REGION,
)

sm = aws_session.client(
    "sagemaker",
    region_name=REGION,
)

status = sm.describe_endpoint(
    EndpointName=ENDPOINT_NAME
)

print("Endpoint:", ENDPOINT_NAME)
print("Status:", status["EndpointStatus"])
```

```python
payload = {
    "age": 42,
    "monthly_spend": 1250000,
    "tenure": 4,
    "support_ticket": 10,
}

response = runtime.invoke_endpoint(
    EndpointName=ENDPOINT_NAME,
    ContentType="application/json",
    Accept="application/json",
    Body=json.dumps(payload).encode("utf-8"),
)

result = response["Body"].read().decode("utf-8")

print("HTTP status:", response["ResponseMetadata"]["HTTPStatusCode"])
print("Prediction:", result)
```
